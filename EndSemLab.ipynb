```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns
# Step 1: Create Dataset
data = {
'Week_ID': ['W1', 'W2', 'W3', 'W4', 'W5'],
'Promo_Budget': [1000, 2000, 1500, 3000, 1200],
'Online_Ads': [500, 800, 600, 1200, 400],
'Footfall': [150, 280, 210, 420, 180]
}
df = pd.DataFrame(data)
```

```python
# Step 2: Scatter Plot
plt.figure(figsize=(6,5))
plt.scatter(df['Promo_Budget'], df['Footfall'])
plt.xlabel("Promo Budget")
plt.ylabel("Footfall")
plt.title("Promo Budget vs Footfall")
plt.show()
```

```python
# Step 3: Multiple Linear Regression
X = df[['Promo_Budget', 'Online_Ads']]
y = df['Footfall']

model = LinearRegression()
model.fit(X, y)
```

```python
# Step 4: Predict Footfall
prediction = model.predict([[2500, 900]])
print("Predicted Footfall:", round(prediction[0],2))
```

```python
# Step 5: Model Coefficients
coefficients = pd.DataFrame({
'Feature': X.columns,
'Coefficient': model.coef_
})
print(coefficients)
```

```python
# Step 6: Pair Plot
sns.pairplot(df[['Promo_Budget', 'Online_Ads', 'Footfall']])
plt.show()
```

```python
# Step 7: Coefficient Bar Chart
plt.figure(figsize=(6,5))
plt.bar(coefficients['Feature'], coefficients['Coefficient'])
plt.xlabel("Features")
plt.ylabel("Coefficient")
plt.title("Model Coefficients")
plt.show()
```
